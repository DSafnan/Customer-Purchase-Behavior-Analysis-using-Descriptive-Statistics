```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/Superstore Marketing Data ')
```

```python
print(df.head())
```

```python
print(df.isnull().sum())
```

###handling missing values by imputation technique

```python
numeric_cols = ['Income', 'MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts',
               'MntSweetProducts', 'MntGoldProds']
for col in numeric_cols:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
print(df.isnull().sum())
```

```python
# For date column, fill with mode
df['Dt_Customer'].fillna(df['Dt_Customer'].mode()[0], inplace=True)
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
#date to datetime format
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'], errors='coerce', format='%m/%d/%Y')
```

```python
df.head()
```

###Feature Engineering

```python
current_year = 2023  # Update based on current year
df['Age'] = current_year - df['Year_Birth']
```

```python
#checing for duplicate
df.duplicated().sum()
```

```python
# Replace 'ID' with the actual column name representing customer ID.
# Assuming the customer ID column is named 'Id', the code should be:
df.drop_duplicates(subset='Id', keep='first', inplace=True)
```

## Task 3 - Probability Distributions 🎲

- **Identify variables that could follow specific probability distributions (e.g., Binomial, Normal). Calculate probabilities and expected values based on these distributions.**



```python
#probability of offer acceptance
acceptance_prob = df['Response'].mean()
print(f"\nProbability of offer acceptance: {acceptance_prob:.2%}")
```

```python
#probability of high wine spending (>500)
high_wine_prob = (df['MntWines'] > 500).mean()
print(f"Probability of spending >$500 on wine: {high_wine_prob:.2%}")
```

## Task 4: Insights and Customer Segmentation 📈

- **Explore relationships between customer characteristics and spending habits. Segment customers based on their behaviors and characteristics.**

```python
df['Customer_Type'] = np.where(df['MntWines'] > 500, 'High Spender',
                              np.where(df['MntWines'] > 100, 'Medium Spender', 'Low Spender'))
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Customer_Type')
plt.title('Customer Segmentation by Spending Level')
plt.show()
```

**Deliverables**:

- **Key insights regarding relationships between variables and distinct customer segments based on behaviors.**
